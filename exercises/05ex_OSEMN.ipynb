4\. load the binary file named *credit_card.dat* and convert the data into the real credit-card number.
Each line correspond to a credit card number.
Each character is composed by 6 bit (even the space) and the last 4 bit are just a padding

**hint**: use the `chr()` function to convert a number to a char

```python
n_bits = 6
cards = [] 

with open('credit_card.dat', 'rb') as f:
    for line in f:
        #Converting each line first in an integer number (base 2) and then into the corrisponding char with chr()
        card = [chr(int(line[i : i + n_bits], 2)) for i in range(0, len(line), n_bits)]
        card = card[:-1]   #Removing the last char as it is '\n'
        
        card_number = ''
        for char in card:
            card_number += char
            
        cards.append(card_number)
        
cards = cards[:-1]  #The last 4 bit of the overall file are padding, so we discard the last value we obtained

print("Credit cards:\n")
for card in cards:
    print(card)
            
        
```

5\. Load the file "user_data.json", filter the data by the "CreditCardType" field equals to "American Express". Than save the data a to CSV.

```python
from json import load, dumps
#user_data = load(open("user_data.json"))
with open('user_data.json') as infile:
    user_data = load(infile)
    with open('user_data.csv', "w") as outfile:
        
        for i in user_data[0].keys():
            outfile.write(i+",")
        outfile.write("\n")
        for l in user_data:
            if l['CreditCardType'] == 'American Express':
                for i in l.values():
                     outfile.write(str(i)+",")
                outfile.write("\n")
                
```
