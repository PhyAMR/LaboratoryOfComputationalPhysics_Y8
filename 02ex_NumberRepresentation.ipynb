2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)


```python
 import random

#Function that generates a random binary number of 32 bits as a string
def generating_bin_number():
    s = ""
    
    for _ in range(32):
        bit = random.choice('01')
        s += bit
    
    return s
bin_number = generating_bin_number()

def bin_to_float(bin_number):
    bias = 127
    sign = int(bin_number[0])
    exponent_bits = bin_number[1:9]
    mantissa_bits = bin_number[9:]

    exponent = sum(int(bit) * 2**(7 - i) for i, bit in enumerate(exponent_bits)) - bias
    mantissa = 1 + sum(int(bit) * 2**(- (i + 1)) for i, bit in enumerate(mantissa_bits))

    float_value = (-1)**sign * mantissa * 2**exponent
    return float_value
float_value = bin_to_float(bin_number)
print(f"Binary: {bin_number} to Float: {float_value}")
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?


```python


def int_f2(N):
  x = np.linspace(-1,1,N)
  f = lambda x: np.sqrt(1-x**2)
  h=2/N 
  return np.sum(h*f(x))
N=100

print(f"The relative error for N={N} is: {100*np.abs(int_f2(N)- np.pi/2)/(np.pi/2)}%")
```

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
import time 

cycle = 0
N=100
while cycle <= 1.0:
  N += 1
  start = time.time()
  print(f"The relative error for N={N} is: {100*np.abs(int_f2(N)- np.pi/2)/(np.pi/2)}%")
  cycle = start - time.time()

```
