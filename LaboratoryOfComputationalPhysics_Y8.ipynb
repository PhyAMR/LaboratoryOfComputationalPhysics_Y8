7\. Prime numbers sieve: compute the prime numbers in the 0-N (N=99 to start with) range with a sieve (mask).
  * Constract a shape (100,) boolean array, the mask
  * Identify the multiples of each number starting from 2 and set accordingly the corresponding mask element
  * Apply the mask to obtain an array of ordered prime numbers
  * Check the performances (timeit); how does it scale with N?
  * Implement the optimization suggested in the [sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)


````python
def primes(N):
    m = np.arange(2, N+1)
    final_mask = np.zeros_like(m, dtype=bool)

    for i in range(2, 10):
        condicion_diferente = (m != i)
        condicion_divisible = (m % i == 0)
        mask_i = condicion_diferente & condicion_divisible
        final_mask = final_mask | mask_i

    primes = m[~final_mask]
    return primes
print(primes(100))
````

````python
def sieve_of_eratosthenes(n):
    if n < 2:
        return []

    # Create a boolean array "prime[0..n]" and initialize
    # all entries it as true. A value in prime[i] will
    # finally be false if i is Not a prime, else true.
    prime = [True] * (n + 1)
    prime[0] = prime[1] = False  # 0 and 1 are not prime numbers

    # We only need to check up to the square root of n
    for p in range(2, int(n**0.5) + 1):
        # If prime[p] is not changed, then it is a prime
        if prime[p]:
            # Update all multiples of p starting from p*p
            for i in range(p * p, n + 1, p):
                prime[i] = False

    # Collect all numbers that are still marked as True
    primes = [i for i in range(2, n + 1) if prime[i]]
    return primes

# Example usage:
print(sieve_of_eratosthenes(30))
````

````python
for n in [100,1000,10000, 100000]:
    print("My way")
    %timeit primes(n)
    print("sieve_of_eratosthenes")
    %timeit sieve_of_eratosthenes(n)
````

8\. Diffusion using random walk

Consider a simple random walk process: at each step in time, a walker jumps right or left (+1 or -1) with equal probability. The goal is to find the typical distance from the origin of a random walker after a given amount of time. 
To do that, let's simulate many walkers and create a 2D array with each walker as a raw and the actual time evolution as columns

  * Take 1000 walkers and let them walk for 200 steps
  * Use randint to create a 2D array of size walkers x steps with values -1 or 1
  * Build the actual walking distances for each walker (i.e. another 2D array "summing on each raw")
  * Take the square of that 2D array (elementwise)
  * Compute the mean of the squared distances at each step (i.e. the mean along the columns)
  * Plot the average distances (sqrt(distance\*\*2)) as a function of time (step)
  
Did you get what you expected?


````python
W = 2*np.random.randint(0,2,size=(1000,200),dtype=int)-1

d = np.zeros_like(W)
for i in range(d.shape[1]):
    d[:,i] = np.sum(W[:,:i], axis=1)

D = d**2
mean_dis = D.mean(axis=0)
````

1\. **Kernel Density Estimate**

Produce a KDE for a given distribution (by hand, not using seaborn!):

* Fill a numpy array, x,  of len(N) (with N=O(100)) with a variable normally distributed, with a given mean a standard deviation
* Fill an histogram in pyplot taking properly care about the aesthetic
   * use a meaningful number of bins
   * set a proper y axis label
   * set proper value of y axis major ticks labels (e.g. you want to display only integer labels)
   * display the histograms as data points with errors (the error being the poisson uncertainty)
* for every element of x, create a gaussian with the mean corresponding the element value and std as a parameter that can be tuned. The std default value should be:
$$ 1.06 * x.std() * x.size ^{-\frac{1}{5.}} $$
you can use the scipy function `stats.norm()` for that.
* In a separate plot (to be placed beside the original histogram), plot all the gaussian functions so obtained
* Sum (with np.sum()) all the gaussian functions and normalize the result such that the integral matches the integral of the original histogram. For that you could use the `scipy.integrate.trapz()` method


````python
from scipy.stats import norm
mean = 0
std = 5
N=100
x = np.random.normal(mean,std,N)
fig, (ax1, ax2) = plt.subplots(1,2, figsize=(12,6))
counts, bins, _ = ax1.hist(x,bins=int(N/5), label="Counts")
ax1.errorbar(x=0.5*(bins[1:] + bins[:-1]),y=counts,yerr=np.sqrt(counts))

dist = []
for i in x:
    h = 1.06*x.std()*x.size**(-1/5)
    disti = norm.pdf(np.linspace(x.min()-3*h, x.max()+3*h, N),loc=i,scale=h)
    dist.append(disti)
g= np.array(dist).sum(axis=0)
hist_int = np.trapezoid(counts, bins[:-1])
g_int = np.trapezoid(g, np.linspace(x.min()-3*h, x.max()+3*h, N))
g_norm = g * (hist_int / g_int)
ax1.plot(np.linspace(x.min()-3*h, x.max()+3*h, N),g_norm)
for gi in dist:
    ax2.plot(np.linspace(x.min()-3*h, x.max()+3*h, N),gi)
plt.show()
````

## Linear Algebra

1\. **PCA on 3D dataset**

Generate a dataset with $x_1\sim N(0,1)$, $x_2 = x_1+N(0,3)$, $x_3=2x_1+x_2$.
Find eigenvectors/eigenvalues of the covariance matrix. Repeat with SVD. Report variance explained. Redefine data in PCA basis and plot 2×3 scatter grid.

````python
from scipy import linalg as la

x1 = np.random.normal(0,1,1000)
x2 = np.random.normal(0,3,1000)+x1
x3 = 2*x1+x2
x4 = np.random.normal(0,0.1,1000)
X = np.column_stack((x1, x2, x3,x4))

# Center the data (crucial for both Covariance and SVD-based PCA)
X_centered = X - np.mean(X, axis=0)

# =====================================================================
# 2. Eigenvectors and Eigenvalues from Covariance Matrix
# =====================================================================
# cov_matrix = (X_centered.T @ X_centered) / (N - 1)
cov_matrix = np.cov(X_centered, rowvar=False)
eigenvalues_cov, eigenvectors_cov = la.eigh(cov_matrix)

# Sort descending
idx_cov = np.argsort(eigenvalues_cov)[::-1]
eigenvalues_cov = eigenvalues_cov[idx_cov]
eigenvectors_cov = eigenvectors_cov[:, idx_cov]
````

````python
U, S, Vt = la.svd(X_centered, full_matrices=False)

# Convert SVD singular values to Covariance eigenvalues: lambda = s^2 / (N - 1)
eigenvalues_svd = (S**2) / (N - 1)
eigenvectors_svd = Vt.T  # Columns are the right singular vectors

# Check alignment / sign corrections (eigenvectors are unique up to a sign flip)
for i in range(X.shape[1]):
    if np.dot(eigenvectors_cov[:, i], eigenvectors_svd[:, i]) < 0:
        eigenvectors_svd[:, i] = -eigenvectors_svd[:, i]

print("--- Method Comparison ---")
print("Covariance Eigenvalues: ", eigenvalues_cov)
print("SVD Eigenvalues:        ", eigenvalues_svd)
print("\nAre the eigenvectors close?", np.allclose(eigenvectors_cov, eigenvectors_svd))

total_variance = np.sum(eigenvalues_cov)
explained_variance_ratio = eigenvalues_cov / total_variance
cumulative_variance_ratio = np.cumsum(explained_variance_ratio)

print("\n--- Variance Analysis ---")
for i, prop in enumerate(explained_variance_ratio):
    print(f"PC{i+1} explains: {prop*100:.2f}% of the variance")
print(f"Cumulative variance: {cumulative_variance_ratio * 100}")

# Determine number of components for >= 99%
n_components = np.argmax(cumulative_variance_ratio >= 0.99) + 1
print(f"To retain >= 99% variability, reduce dimensions to: {n_components} components")
````

````python
from itertools import combinations

X_pca = X_centered @ eigenvectors_cov

# =====================================================================
# 6. Plotting the Results
# =====================================================================
pairs = [(i,j) for i,j in combinations(range(X.shape[1]),2)]
print(pairs)
fig, axes = plt.subplots(2, len(pairs), figsize=(15, 10))

# Row 1: Original Coordinates
for col, (i, j) in enumerate(pairs):
    axes[0, col].scatter(X[:, i], X[:, j], alpha=0.5, color='b', edgecolors='none')
    axes[0, col].set_title(f"Original: Feature {i} vs {j}")
    axes[0, col].set_xlabel(f"x_{i+1}")
    axes[0, col].set_ylabel(f"x_{j+1}")
    axes[0, col].grid(True, linestyle='--', alpha=0.6)

# Row 2: PCA Coordinates
for col, (i, j) in enumerate(pairs):
    axes[1, col].scatter(X_pca[:, i], X_pca[:, j], alpha=0.5, color='r' if i==2 or j==2 else 'crimson', edgecolors='none')
    axes[1, col].set_title(f"PCA Basis: PC {i+1} vs PC {j+1}")
    axes[1, col].set_xlabel(f"PC {i+1}")
    axes[1, col].set_ylabel(f"PC {j+1}")
    axes[1, col].grid(True, linestyle='--', alpha=0.6)
    
    # Highlight the zero-variance dimension if it exists
    if j == 2: 
        axes[1, col].set_ylim(-1, 1) # Zoom in to show it is flat

plt.tight_layout()
plt.show()
````

1\. **Maximum wind speed prediction at the Sprogø station**

The exercise goal is to predict the maximum wind speed occurring every 50 years even if no measure exists for such a period. The available data are only measured over 21 years at the Sprogø meteorological station located in Denmark. 

The annual maxima are supposed to fit a normal probability density function. However such function is not going to be estimated because it gives a probability from a wind speed maxima. Finding the maximum wind speed occurring every 50 years requires the opposite approach, the result needs to be found from a defined probability. That is the quantile function role and the exercise goal will be to find it. In the current model, it is supposed that the maximum wind speed occurring every 50 years is defined as the upper 2% quantile.

By definition, the quantile function is the inverse of the cumulative distribution function. The latter describes the probability distribution of an annual maxima. In the exercise, the cumulative probability $p_i$ for a given year i is defined as $p_i = i/(N+1)$ with $N = 21$, the number of measured years. Thus it will be possible to calculate the cumulative probability of every measured wind speed maxima. From those experimental points, the scipy.interpolate module will be very useful for fitting the quantile function. Finally the 50 years maxima is going to be evaluated from the cumulative probability of the 2% quantile.

Practically, load the dataset:

```python
import numpy as np
max_speeds = np.load('max-speeds.npy')
years_nb = max_speeds.shape[0]
```

Compute then the cumulative probability $p_i$ (`cprob`) and sort the maximum speeds from the data. Use then the  UnivariateSpline from scipy.interpolate to define a quantile function and thus estimate the probabilities.

In the current model, the maximum wind speed occurring every 50 years is defined as the upper 2% quantile. As a result, the cumulative probability value will be:

```python
fifty_prob = 1. - 0.02
```

So the storm wind speed occurring every 50 years can be guessed as:

``` python
fifty_wind = quantile_func(fifty_prob)
```


````python
from scipy.interpolate import UnivariateSpline
max_speeds = np.load('max-speeds.npy')
years_nb = max_speeds.shape[0]
nprob = np.linspace(0,0.99, 1000)
cprob = (np.arange(years_nb, dtype = np.float32) + 1) / (years_nb + 1)

sorted_speed = np.sort(max_speeds)

quantile_func = UnivariateSpline(cprob,sorted_speed)

fitted_data = quantile_func(nprob)
fifty_prob = 1. - 0.02

fifty_wind = quantile_func(fifty_prob)

plt.scatter(sorted_speed, cprob, label = 'Data')
plt.xlabel('Max speeds')
plt.plot(fitted_data, nprob, linestyle = '--', label = 'Fit', c = 'green')
plt.axvline(fifty_wind)
plt.ylabel('Probability')
plt.legend()
````
