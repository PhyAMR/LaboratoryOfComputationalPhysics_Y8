3\. Load the saved txt file of point 2 and convert it to a csv file (by hand)

```python


with open('data.txt') as infile:
    with open('data.csv', "w") as outfile:
        for l in infile:
            line = str(l).replace(" ", ",")
            outfile.write(line)
```
